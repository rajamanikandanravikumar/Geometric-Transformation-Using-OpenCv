# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
height,width,dim = image.shape
image.shape
```

```python
M1 = np.float32([
    [1, 0.5, 0],
    [0, 1, 0]
])

h_image = cv2.warpAffine(image, M1, (1400, height))
```

```python
M2 = np.float32([
    [1, 0, 0],
    [0.5, 1, 0]
])

v_image = cv2.warpAffine(image, M2, (width, 1200))
```

```python
plt.subplot(1,3,1)
plt.title("Original Image")
plt.imshow(image)
plt.axis('off')
plt.show()

plt.subplot(1,3,2)
plt.title("Horizontal Image")
plt.imshow(h_image)
plt.axis('off')
plt.show()

plt.subplot(1,3,3)
plt.title("Vertical Image")
plt.imshow(v_image)
plt.axis('off')
plt.show()
```

# **Image Reflection**

Write a Python program using OpenCV to perform image reflection on the **given image**.

The program should:

1. Generate the following reflected versions of the image:
   - **Horizontal Reflection** (flip left ↔ right)  
   - **Vertical Reflection** (flip top ↔ bottom)  
   - **Both-axis Reflection**  
2. Use `cv2.flip` with appropriate flip codes.
3. Display the following using a **2 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Reflected Image**  
   - **Vertically Reflected Image**  
   - **Both-axis Reflected Image**


```python
horizontal_flip = cv2.flip(image,1)
```

```python
vertical_flip = cv2.flip(image,0)
```

```python
axis_reflect = cv2.flip(image,-1)
```

```python
plt.subplot(2,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(horizontal_flip)
plt.title("Horizontal Reflection")
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(vertical_flip)
plt.title("Vertical Reflection")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(axis_reflect)
plt.title("Both-axis Reflection")
plt.axis('off')

plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
M1 = cv2.getRotationMatrix2D((width//2,height//2),45,1)
rotate45 = cv2.warpAffine(image, M1, (width, height))
```

```python
M2 = cv2.getRotationMatrix2D((width//2, height//2), 90, 1)
rotate90 = cv2.warpAffine(image, M2, (width, height))
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(rotate45)
plt.title("45° Rotated Image")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(rotate90)
plt.title("90° Rotated Image")
plt.axis('off')

plt.show()
```
